```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employees.csv")

print("First five records:")
print(df.head())

salary = df["Salary"]

print("\nMean   :", round(salary.mean(), 2))
print("Median :", salary.median())
print("Mode   :", salary.mode().tolist())

print("\nMinimum salary:", salary.min())
print("Maximum salary:", salary.max())

print("\nVariance           :", round(salary.var(), 2))
print("Standard Deviation :", round(salary.std(), 2))

plt.figure(figsize=(8, 5))
plt.hist(salary, bins=8, color="skyblue", edgecolor="black")
plt.axvline(salary.mean(), color="red", linestyle="--", label="Mean")
plt.axvline(salary.median(), color="green", linestyle="-.", label="Median")
plt.title("Histogram of Employee Salary")
plt.xlabel("Salary (Rs.)")
plt.ylabel("Number of employees")
plt.legend()
plt.show()
```
