```python
import pandas as pd
from scipy import stats

df = pd.read_csv("StudentsPerformance.csv")

male = df[df["gender"] == "male"]["math score"]
female = df[df["gender"] == "female"]["math score"]

print("Male   : n =", len(male), " mean =", round(male.mean(), 2), " std =", round(male.std(), 2))
print("Female : n =", len(female), " mean =", round(female.mean(), 2), " std =", round(female.std(), 2))

t_stat, p_value = stats.ttest_ind(male, female)
print("\nt-statistic =", round(t_stat, 4))
print("p-value     =", p_value)

alpha = 0.05
if p_value < alpha:
    print("\nReject H0: there is a significant difference in the average Mathematics scores.")
else:
    print("\nFail to reject H0: no significant difference in the average Mathematics scores.")
```
