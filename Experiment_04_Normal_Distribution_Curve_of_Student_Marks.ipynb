```python
import numpy as np
import matplotlib.pyplot as plt

students = {
    "Student_ID": ["S101", "S102", "S103", "S104", "S105", "S106", "S107", "S108", "S109", "S110",
                   "S111", "S112", "S113", "S114", "S115", "S116", "S117", "S118", "S119", "S120"],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan", "Neha", "Rohan", "Pooja", "Vivek", "Anjali",
             "Akash", "Riya", "Nikhil", "Komal", "Arjun", "Meera", "Sagar", "Aarti", "Rohit", "Kavya"],
    "Marks": [78, 85, 72, 91, 88, 76, 69, 84, 81, 90, 73, 87, 79, 82, 95, 80, 75, 89, 77, 93],
}

marks = np.array(students["Marks"])

mean = np.mean(marks)

std = np.std(marks)
print("Mean               =", round(mean, 2))
print("Standard deviation =", round(std, 2))

plt.figure(figsize=(8, 5))
plt.hist(marks, bins=7, density=True, color="lightblue", edgecolor="black", label="Marks histogram")

x = np.linspace(mean - 4 * std, mean + 4 * std, 200)

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2)

plt.plot(x, y, color="red", linewidth=2, label="Normal curve")

plt.title("Normal Distribution Curve of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Probability density")
plt.legend()
plt.show()
```
