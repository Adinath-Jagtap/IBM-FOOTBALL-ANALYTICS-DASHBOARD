```python
import pandas as pd

df = pd.read_csv("students.csv")

print("Complete dataset:")
print(df)

simple = df.sample(n=6, random_state=42)

stratified = df.groupby("Department").sample(n=2, random_state=42)

cluster = df[df["Department"] == "COMP"]

print("\nSimple Random Sample (6 students):")
print(simple)
print("\nStratified Sample (2 per department):")
print(stratified)
print("\nCluster Sample (COMP department):")
print(cluster)

comparison = pd.DataFrame({
    "Method": ["Simple Random", "Stratified", "Cluster"],
    "Sample size": [len(simple), len(stratified), len(cluster)],
    "Departments covered": [simple["Department"].nunique(),
                            stratified["Department"].nunique(),
                            cluster["Department"].nunique()],
    "Mean marks": [round(simple["Marks"].mean(), 2),
                   round(stratified["Marks"].mean(), 2),
                   round(cluster["Marks"].mean(), 2)],
})
print("\nComparison:")
print(comparison)
print("\nPopulation mean marks:", round(df["Marks"].mean(), 2))
```
