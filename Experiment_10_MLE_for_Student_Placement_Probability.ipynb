```python
csv = """Student,Placement
S1,1
S2,1
S3,0
S4,1
S5,0
S6,1
S7,1
S8,0
S9,1
S10,1
S11,0
S12,1
S13,1
S14,0
S15,1
S16,0
S17,1
S18,1
S19,0
S20,1
"""
with open("placement.csv", "w") as f:
    f.write(csv)
print("Created", "placement.csv")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("placement.csv")
data = df["Placement"].values
n = len(data)
k = data.sum()

p_hat = k / n
print("Total students      :", n)
print("Placed students    :", k)
print("MLE of placement probability p =", p_hat)

p_values = np.linspace(0.01, 0.99, 99)
log_lik = k * np.log(p_values) + (n - k) * np.log(1 - p_values)
best_p = p_values[np.argmax(log_lik)]
print("p that maximises the log-likelihood =", round(best_p, 2))

plt.figure(figsize=(8, 5))
plt.plot(p_values, log_lik, color="blue")
plt.axvline(p_hat, color="red", linestyle="--", label="MLE p = " + str(p_hat))
plt.title("Log-Likelihood of the Bernoulli Model")
plt.xlabel("p")
plt.ylabel("Log-likelihood")
plt.legend()
plt.show()
```
