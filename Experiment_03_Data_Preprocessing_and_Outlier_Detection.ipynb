```python
csv = """Product_ID,Category,Price,Quantity_Sold
P101,Electronics,25000.0,120.0
P102,Grocery,1500.0,250.0
P103,Electronics,,180.0
P104,Clothing,3500.0,
P105,Grocery,1800.0,300.0
P106,Clothing,4000.0,200.0
P107,Electronics,150000.0,150.0
"""
with open("product_sales.csv", "w") as f:
    f.write(csv)
print("Created", "product_sales.csv")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

data = {
    "Product_ID": ["P101", "P102", "P103", "P104", "P105", "P106", "P107"],
    "Category": ["Electronics", "Grocery", "Electronics", "Clothing", "Grocery", "Clothing", "Electronics"],
    "Price": [25000, 1500, np.nan, 3500, 1800, 4000, 150000],
    "Quantity_Sold": [120, 250, 180, np.nan, 300, 200, 150],
}
df = pd.DataFrame(data)

print("Original dataset:")
print(df)

print("\nMissing values:")
print(df.isnull().sum())

df["Price"] = df["Price"].fillna(df["Price"].mean())

df["Quantity_Sold"] = df["Quantity_Sold"].fillna(df["Quantity_Sold"].median())

le = LabelEncoder()
df["Category_Encoded"] = le.fit_transform(df["Category"])

print("\nAfter handling missing values and encoding:")
print(df)

Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)
IQR = Q3 - Q1
print("\nQ1  =", Q1, " Q3 =", Q3, " IQR =", IQR)

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print("Lower Limit =", lower, " Upper Limit =", upper)

outliers = df[(df["Price"] < lower) | (df["Price"] > upper)]
print("\nOutlier(s) in Price:")
print(outliers)

print("\nFinal processed dataset:")
print(df)
```
