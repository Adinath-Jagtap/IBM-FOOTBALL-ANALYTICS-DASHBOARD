```python
import pandas as pd
from scipy.stats import chi2_contingency

df = pd.read_csv("Titanic.csv")

table = pd.crosstab(df["Sex"], df["Survived"])
print("Observed frequencies:")
print(table)

chi2, p, dof, expected = chi2_contingency(table)
print("\nChi-square statistic =", round(chi2, 4))
print("Degrees of freedom   =", dof)
print("p-value              =", p)
print("\nExpected frequencies:")
print(pd.DataFrame(expected, index=table.index, columns=table.columns).round(2))

alpha = 0.05
if p < alpha:
    print("\nReject H0: survival is associated with gender.")
else:
    print("\nFail to reject H0: survival is independent of gender.")
```
